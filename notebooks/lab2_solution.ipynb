```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
DATA_PATH = next((p for p in [Path('../data/tayseer_services.csv'),Path('../data/dashboard_summary.csv')] if p.exists()), Path('../data/dashboard_summary.csv'))
df = pd.read_csv(DATA_PATH, parse_dates=['month'])
df.head()
```

## Task classification
1. Lowest three regions → ranking/comparison → sorted bar.
2. Riyadh over last year → time trend → line.
3. CSAT vs transaction cost → relationship → scatter.
4. Completion time → distribution → histogram.
5. Highest-adoption regions → ranking first; map only if spatial pattern is the question.

A common-axis bar is easier to compare across 13 regions than repeated pie charts. For channel composition across many regions, use a 100% stacked bar rather than 13 separate pies.

```python
latest_month=df['month'].max()
latest=df[df.month==latest_month].sort_values('digital_adoption_pct')
bottom3=latest.head(3)
ax=bottom3.plot.barh(x='region',y='digital_adoption_pct',legend=False,title='Three lowest-adoption regions')
ax.axvline(65,linestyle='--')
ax.set_xlabel('Digital adoption (%)')
plt.show()
```

```python
riyadh=df[df.region=='Riyadh'].sort_values('month').tail(12)
ax=riyadh.plot(x='month',y='digital_adoption_pct',marker='o',legend=False,title='Riyadh adoption — last 12 months')
ax.axhline(65,linestyle='--')
plt.show()
```

## Reflection
The chart type follows the analytical task, not the column type. Geography does not automatically imply a choropleth. The goal is a perceptually accurate comparison with the least unnecessary decoding.
