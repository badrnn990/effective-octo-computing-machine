# Lab 4 — Interactive Tayseer Dashboard
SDA-DSC-112 · Tayseer

Completed solution demonstrating interactive Plotly views and the transition from charts to decision support. The production dashboard is implemented in `../app.py`.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
DATA_PATH=next((p for p in [Path('../data/tayseer_services.csv'),Path('../data/dashboard_summary.csv')] if p.exists()),Path('../data/dashboard_summary.csv'))
df=pd.read_csv(DATA_PATH,parse_dates=['month'])
TARGET=65.0
latest=df[df.month==df.month.max()].sort_values('digital_adoption_pct')
```

```python
fig=px.bar(latest,x='digital_adoption_pct',y='region',orientation='h',hover_data=['unique_users'])
fig.add_vline(x=TARGET,line_dash='dash')
fig.update_layout(title='Latest regional adoption — interactive')
fig.show()
```

```python
fig=go.Figure()
regions=sorted(df.region.unique())
for i,r in enumerate(regions):
    sub=df[df.region==r].sort_values('month')
    fig.add_trace(go.Scatter(x=sub.month,y=sub.digital_adoption_pct,name=r,visible=(i==0)))
buttons=[]
for i,r in enumerate(regions):
    visible=[False]*len(regions); visible[i]=True
    buttons.append(dict(label=r,method='update',args=[{'visible':visible},{'title':f'{r} adoption trend'}]))
fig.update_layout(updatemenus=[dict(buttons=buttons)],title=f'{regions[0]} adoption trend')
fig.show()
```

## Reflection
Hover and zoom let a reader inspect context without requesting new code. A dropdown turns 13 separate region views into self-service exploration. A single executive screen preserves context between headline KPI, trend and regional gap. The final `app.py` extends this idea with visible filter state, measure switching, latest-month regional comparison and supporting KPIs.
